```python

import sys
import os, glob, random, shutil, subprocess, zipfile, json, math, time
from pathlib import Path

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q",
                       "segmentation-models-pytorch", "timm",
                       "opencv-python-headless", "kagglehub"])

import numpy as np
import pandas as pd
import cv2
from PIL import Image
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torchvision
from sklearn.model_selection import train_test_split
from sklearn.metrics import (roc_auc_score, accuracy_score, balanced_accuracy_score,
                             confusion_matrix, classification_report, roc_curve)

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED) if torch.cuda.is_available() else None
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
if device.type != "cuda":
    print("WARNING: GPU not detected. In Colab choose Runtime > Change runtime type > T4 GPU.")

BASE = Path("/content/glaucoma_project")
DATA = BASE / "data"
OUT = BASE / "outputs"
REFUGE_DIR = DATA / "refuge"
OUT.mkdir(parents=True, exist_ok=True)
DATA.mkdir(parents=True, exist_ok=True)

import requests
rar_path = REFUGE_DIR / "Refuge.rar"
REFUGE_DIR.mkdir(parents=True, exist_ok=True)

if not (REFUGE_DIR / "Refuge" / "Training400").exists():
    meta = requests.get("https://api.figshare.com/v2/articles/26049574", timeout=60).json()
    matches = [f for f in meta.get("files", []) if f.get("name") == "Refuge.rar"]
    assert matches, "Could not find Refuge.rar in the Figshare article. Check the REFUGE download source."
    if not rar_path.exists():
        with requests.get(matches[0]["download_url"], stream=True, timeout=180) as r:
            r.raise_for_status()
            with open(rar_path, "wb") as f:
                for chunk in r.iter_content(1024 * 1024):
                    if chunk: f.write(chunk)
    if shutil.which("unrar") is None:
        subprocess.run("apt-get -qq update && apt-get -qq install -y unrar", shell=True, check=True)
    subprocess.run(["unrar", "x", "-idq", "-o+", str(rar_path), str(REFUGE_DIR) + "/"], check=True)

R = REFUGE_DIR / "Refuge"
assert (R / "Training400").exists(), f"REFUGE folder not found at {R / 'Training400'}"

records = []
for folder, label in [("Glaucoma", 1), ("Non-Glaucoma", 0)]:
    img_dir = R / "Training400" / folder
    mask_dir = R / "Annotation-Training400" / "Disc_Cup_Masks" / folder
    for img in sorted(img_dir.glob("*.jpg")):
        mask = mask_dir / (img.stem + ".bmp")
        if mask.exists():
            records.append({"image": str(img), "mask": str(mask), "label": label, "split": "train"})

test_img_dir = R / "Test400"
test_mask_root = R / "REFUGE-Test-GT" / "Disc_Cup_Masks"
for folder, label in [("G", 1), ("N", 0)]:
    for mask in sorted((test_mask_root / folder).glob("*.bmp")):
        img = test_img_dir / (mask.stem + ".jpg")
        if img.exists():
            records.append({"image": str(img), "mask": str(mask), "label": label, "split": "test"})

df = pd.DataFrame(records)
assert len(df[df.split == "train"]) > 300, "Too few REFUGE training images found; check extracted folder structure."
assert len(df[df.split == "test"]) > 100, "Too few REFUGE test images found; check extracted folder structure."
print(df.groupby(["split", "label"]).size().rename("images"))
print("Total labelled images:", len(df))

train_all = df[df.split == "train"].reset_index(drop=True)
test_df = df[df.split == "test"].reset_index(drop=True)

patient_map_path = DATA / "REFUGE_patient_mapping.csv"
if patient_map_path.exists():
    patient_map = pd.read_csv(patient_map_path)
    if not {"image", "patient_id"}.issubset(patient_map.columns):
        raise ValueError("REFUGE_patient_mapping.csv must contain image,patient_id.")
    patient_map["image_name"] = patient_map["image"].map(lambda x: Path(str(x)).name)
    lookup = dict(zip(patient_map["image_name"], patient_map["patient_id"]))
    train_all["patient_id"] = train_all["image"].map(lambda x: lookup.get(Path(x).name))
    if train_all["patient_id"].isna().any():
        missing = train_all.loc[train_all["patient_id"].isna(), "image"].head().tolist()
        raise ValueError(f"Patient mapping missing REFUGE images, examples: {missing}")
    if train_all.groupby("patient_id")["label"].nunique().gt(1).any():
        raise ValueError("A patient has conflicting labels in REFUGE_patient_mapping.csv.")
    from sklearn.model_selection import StratifiedGroupKFold
    splitter = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
    train_idx, val_idx = next(splitter.split(
        np.arange(len(train_all)), train_all.label.values, groups=train_all.patient_id.values
    ))
    assert set(train_all.iloc[train_idx].patient_id).isdisjoint(
        set(train_all.iloc[val_idx].patient_id)
    ), "Patient leakage detected between train and validation."
    print("Using patient-level stratified split from REFUGE_patient_mapping.csv.")
else:
    print("WARNING: No authoritative REFUGE patient mapping supplied. Using image-level "
          "stratified split; verify one-image-per-patient in official metadata before "
          "claiming patient-level separation.")
    train_idx, val_idx = train_test_split(
        np.arange(len(train_all)), test_size=0.20, random_state=SEED,
        stratify=train_all.label.values
    )

train_df = train_all.iloc[train_idx].reset_index(drop=True)
val_df = train_all.iloc[val_idx].reset_index(drop=True)
for name, frame in [("Train", train_df), ("Validation", val_df), ("Test", test_df)]:
    assert frame.label.nunique() == 2, f"{name} must contain both classes"
print(f"Train={len(train_df)} | validation={len(val_df)} | untouched official test={len(test_df)}")
print("Glaucoma fraction:", round(train_df.label.mean(), 3))


# ============================================================
# AUTOMATIC DOWNLOAD OF EXTERNAL DATASETS USING KAGGLEHUB
# ============================================================

import kagglehub

EXTERNAL_ROOT = DATA / "external"
EXTERNAL_ROOT.mkdir(parents=True, exist_ok=True)

def download_dataset(handle, name):
    destination = EXTERNAL_ROOT / name
    destination.mkdir(parents=True, exist_ok=True)

    print(f"\nDownloading {name}...")
    path = kagglehub.dataset_download(
        handle,
        output_dir=str(destination)
    )
    print(f"Downloaded {name}: {path}")
    return Path(path)


# DRISHTI-GS: segmentation images and annotations
DRISHTI_PATH = download_dataset(
    "lokeshsaipureddi/drishtigs-retina-dataset-for-onh-segmentation",
    "DRISHTI-GS"
)

# DRISHTI-GS: helper classification labels
DRISHTI_LABEL_PATH = download_dataset(
    "chetanpediredla/glaucoma-dataset",
    "DRISHTI-LABELS"
)

# RIM-ONE DL: external classification dataset
RIM_PATH = download_dataset(
    "orvile/rim-one-retinal-dataset-for-assessing-glaucoma",
    "RIM-ONE-V3"
)

# REFUGE + ORIGA bundle
BUNDLE_PATH = download_dataset(
    "arnavjain1/glaucoma-datasets",
    "GLAUCOMA-BUNDLE"
)

# ACRIMA: external classification dataset
try:
    ACRIMA_PATH = download_dataset(
        "orvile/acrima-glaucoma-assessment-using-fundus-images",
        "ACRIMA"
    )
except Exception as e:
    print("Primary ACRIMA source failed:", e)
    ACRIMA_PATH = download_dataset(
        "toaharahmanratul/acrima-dataset",
        "ACRIMA"
    )


# Locate ORIGA inside the combined bundle
def find_dataset_folder(root, folder_name):
    matches = [
        p for p in Path(root).rglob("*")
        if p.is_dir() and p.name.lower() == folder_name.lower()
    ]
    if not matches:
        raise FileNotFoundError(
            f"{folder_name} folder not found inside {root}"
        )
    return matches[0]


ORIGA_PATH = find_dataset_folder(BUNDLE_PATH, "ORIGA")


IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

LABEL_NAMES = {
    "normal": 0, "non-glaucoma": 0, "nonglaucoma": 0, "healthy": 0, "non_glaucoma": 0,
    "glaucoma": 1, "glaucomatous": 1
}

EXTERNAL_DATASETS = {
    "DRISHTI-GS": DRISHTI_PATH,
    "RIM-ONE-V3": RIM_PATH,
    "ORIGA": ORIGA_PATH,
    "ACRIMA": ACRIMA_PATH,
}


# ============================================================
# EXTERNAL DATASET RECORDS AND MANIFEST PROCESSING
# ============================================================

def external_records(dataset_name, root):
    if not root.exists():
        return []

    manifest = EXTERNAL_ROOT / f"{dataset_name}_manifest.csv"
    rows = []

    if manifest.exists():
        mf = pd.read_csv(manifest)

        if not {"image", "label"}.issubset(mf.columns):
            raise ValueError(f"{manifest} must contain image,label columns.")

        for _, rr in mf.iterrows():
            if pd.isna(rr["image"]) or not str(rr["image"]).strip():
                raise ValueError(f"Empty image path in {manifest}")

            p = (root / str(rr["image"])).resolve()

            if not p.is_relative_to(root.resolve()):
                raise ValueError(f"Manifest path escapes dataset directory: {rr['image']}")

            if not p.is_file():
                raise FileNotFoundError(f"Missing image listed in {manifest}: {p}")

            if pd.isna(rr["label"]):
                raise ValueError(f"Missing label for {rr['image']} in {manifest}")

            try:
                label_value = int(rr["label"])
            except (TypeError, ValueError, OverflowError):
                raise ValueError(f"Invalid label for {rr['image']}: {rr['label']}")

            if label_value not in (0, 1) or float(rr["label"]) != label_value:
                raise ValueError(f"Label must be 0 or 1 for {rr['image']}: {rr['label']}")

            rows.append({
                "image": str(p),
                "label": label_value,
                "dataset": dataset_name
            })

    else:
        for p in sorted(root.rglob("*")):
            if p.is_file() and p.suffix.lower() in IMAGE_EXTS:
                label = None

                for part in reversed(p.relative_to(root).parts[:-1]):
                    key = "".join(
                        ch for ch in part.lower()
                        if ch.isalnum() or ch == "-"
                    )
                    if key in LABEL_NAMES:
                        label = LABEL_NAMES[key]
                        break

                if label is not None:
                    rows.append({
                        "image": str(p),
                        "label": label,
                        "dataset": dataset_name
                    })

    unique = {r["image"]: r for r in rows}
    rows = list(unique.values())

    if rows:
        counts = pd.Series([r["label"] for r in rows]).value_counts().to_dict()
        print(f"{dataset_name}: {len(rows)} labelled images; class counts {counts}")

        if set(counts) != {0, 1}:
            raise ValueError(
                f"{dataset_name} must contain both classes (0=Normal, 1=Glaucoma) "
                f"before evaluation. Found class counts: {counts}"
            )
    else:
        print(f"{dataset_name}: no usable labels found. Add class folders or a manifest.")

    return rows


external_frames = {}

for ds_name, ds_root in EXTERNAL_DATASETS.items():
    recs = external_records(ds_name, ds_root)
    if recs:
        external_frames[ds_name] = pd.DataFrame(recs)


# Create manifest templates for datasets without verified labels
for ds_name, ds_root in EXTERNAL_DATASETS.items():
    manifest = EXTERNAL_ROOT / f"{ds_name}_manifest.csv"

    if not manifest.exists() and ds_root.exists():
        image_paths = [
            p for p in sorted(ds_root.rglob("*"))
            if p.is_file() and p.suffix.lower() in IMAGE_EXTS
        ]

        if image_paths:
            template = pd.DataFrame({
                "image": [str(p.relative_to(ds_root)) for p in image_paths],
                "label": [""] * len(image_paths)
            })

            template.to_csv(
                EXTERNAL_ROOT / f"{ds_name}_manifest_TEMPLATE.csv",
                index=False
            )

            print(
                f"Created label template for {ds_name}; "
                f"fill labels from official metadata before evaluation."
            )


# ============================================================
# REFUGE MASK VALIDATION
# ============================================================

mask_values_seen = set()

for _, row in train_df.head(5).iterrows():
    mask_check = cv2.imread(row["mask"], cv2.IMREAD_GRAYSCALE)

    if mask_check is None:
        raise RuntimeError(f"Could not read annotation mask: {row['mask']}")

    mask_values_seen.update(np.unique(mask_check).tolist())

    print(
        Path(row["image"]).name,
        "mask values/counts:", np.unique(mask_check, return_counts=True),
        "cup pixels:", int(np.sum(mask_check < 64)),
        "disc pixels:", int(np.sum(mask_check < 192))
    )

allowed_values = {0, 128, 255}

if not mask_values_seen.issubset(allowed_values):
    raise ValueError(
        f"Unexpected REFUGE mask values: {sorted(mask_values_seen)}. "
        "Inspect the annotation format before training."
    )

if not {0, 255}.issubset(mask_values_seen):
    raise ValueError(
        "Mask encoding requires further inspection: expected to observe cup value 0 "
        "and background value 255 across the inspected masks."
    )


# Visualize REFUGE images and annotations
fig, axes = plt.subplots(3, 3, figsize=(10, 9))

for i, (_, row) in enumerate(train_df.head(3).iterrows()):
    image_check = cv2.cvtColor(
        cv2.imread(row["image"]),
        cv2.COLOR_BGR2RGB
    )

    mask_check = cv2.imread(
        row["mask"],
        cv2.IMREAD_GRAYSCALE
    )

    axes[i, 0].imshow(image_check)
    axes[i, 0].set_title("Fundus")

    axes[i, 1].imshow(mask_check, cmap="gray")
    axes[i, 1].set_title("Original mask")

    axes[i, 2].imshow(image_check)
    axes[i, 2].imshow(mask_check < 192, alpha=0.35, cmap="Greens")
    axes[i, 2].imshow(mask_check < 64, alpha=0.45, cmap="Reds")
    axes[i, 2].set_title("Disc (green), cup (red)")

    for ax in axes[i]:
        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

import segmentation_models_pytorch as smp

SEG_SIZE = 384
SEG_BATCH = 6
SEG_EPOCHS = 25
SEG_WORKERS = 2
SEG_BEST = OUT / "unet_resnet34_best.pth"

IMAGENET_MEAN = np.array([0.485, 0.456, 0.406], np.float32)
IMAGENET_STD = np.array([0.229, 0.224, 0.225], np.float32)

def read_fundus(path, size):
    bgr = cv2.imread(path)
    if bgr is None: raise ValueError(f"Cannot read image: {path}")
    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    return cv2.resize(rgb, (size, size), interpolation=cv2.INTER_AREA)

def read_disc_cup(mask_path, size):
    m = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    if m is None: raise ValueError(f"Cannot read mask: {mask_path}")
    disc = (m < 192).astype(np.uint8)
    cup = (m < 64).astype(np.uint8)
    disc = cv2.resize(disc, (size, size), interpolation=cv2.INTER_NEAREST)
    cup = cv2.resize(cup, (size, size), interpolation=cv2.INTER_NEAREST)
    cup = cup * disc
    return np.stack([disc, cup], axis=0)

class FundusSegDataset(Dataset):
    def __init__(self, frame, augment=False):
        self.frame = frame.reset_index(drop=True)
        self.augment = augment
    def __len__(self): return len(self.frame)
    def __getitem__(self, idx):
        r = self.frame.iloc[idx]
        x = read_fundus(r.image, SEG_SIZE)
        y = read_disc_cup(r["mask"], SEG_SIZE)
        if self.augment:
            if random.random() < 0.5:
                x = np.ascontiguousarray(x[:, ::-1]); y = np.ascontiguousarray(y[:, :, ::-1])
            if random.random() < 0.5:
                alpha = random.uniform(0.9, 1.1); beta = random.uniform(-10, 10)
                x = np.clip(x.astype(np.float32) * alpha + beta, 0, 255).astype(np.uint8)
        x = x.astype(np.float32) / 255.0
        x = (x - IMAGENET_MEAN) / IMAGENET_STD
        return torch.from_numpy(x.transpose(2,0,1).copy()), torch.from_numpy(y.astype(np.float32))

seg_train_ds = FundusSegDataset(train_df, augment=True)
seg_val_ds = FundusSegDataset(val_df, augment=False)
seg_test_ds = FundusSegDataset(test_df, augment=False)
seg_train_loader = DataLoader(seg_train_ds, batch_size=SEG_BATCH, shuffle=True, num_workers=SEG_WORKERS, pin_memory=True)
seg_val_loader = DataLoader(seg_val_ds, batch_size=SEG_BATCH, shuffle=False, num_workers=SEG_WORKERS, pin_memory=True)
seg_test_loader = DataLoader(seg_test_ds, batch_size=SEG_BATCH, shuffle=False, num_workers=SEG_WORKERS, pin_memory=True)

seg_model = smp.Unet(
    encoder_name="resnet34", encoder_weights="imagenet",
    in_channels=3, classes=2, activation=None
).to(device)

bce_loss = nn.BCEWithLogitsLoss()
dice_loss = smp.losses.DiceLoss(mode="multilabel", from_logits=True)

def seg_loss_fn(logits, targets):
    return 0.5 * bce_loss(logits, targets) + 0.5 * dice_loss(logits, targets)

def dice_per_channel(pred, target, eps=1e-7):
    vals = []
    for c in range(2):
        p = pred[:,c].reshape(pred.shape[0], -1).bool()
        t = target[:,c].reshape(target.shape[0], -1).bool()
        inter = (p & t).sum(1).float()
        vals.append(((2*inter + eps) / (p.sum(1) + t.sum(1) + eps)).mean().item())
    return vals

@torch.no_grad()
def evaluate_seg(model, loader):
    model.eval()
    all_dice = []

    for x, y in loader:
        x = x.to(device)
        y = y.to(device)
        logits = model(x)
        pred = torch.sigmoid(logits) >= 0.5
        pred[:, 1] = pred[:, 1] & pred[:, 0]

        batch_scores = []
        for c in range(2):
            p = pred[:, c].flatten(1)
            t = y[:, c].flatten(1).bool()
            inter = (p & t).sum(dim=1).float()
            pred_area = p.sum(dim=1).float()
            target_area = t.sum(dim=1).float()
            dice = (2 * inter + 1e-7) / (pred_area + target_area + 1e-7)
            iou = (inter + 1e-7) / (pred_area + target_area - inter + 1e-7)
            batch_scores.append(torch.stack([dice, iou], dim=1).cpu().numpy())

        all_dice.append(np.stack(batch_scores, axis=1))

    # Shape: images x classes x (Dice, IoU); average per image, not per batch.
    scores = np.concatenate(all_dice, axis=0).mean(axis=0)
    return scores[:, 0], scores[:, 1]

seg_opt = torch.optim.AdamW(seg_model.parameters(), lr=2e-4, weight_decay=1e-4)
seg_sched = torch.optim.lr_scheduler.ReduceLROnPlateau(seg_opt, mode="max", factor=0.5, patience=3)
seg_scaler = torch.amp.GradScaler("cuda", enabled=(device.type == "cuda"))
best_seg_score, patience, no_improve = -1, 7, 0
seg_history = []

for epoch in range(1, SEG_EPOCHS + 1):
    seg_model.train(); running = 0.0
    for x,y in seg_train_loader:
        x,y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        seg_opt.zero_grad(set_to_none=True)
        with torch.autocast(device_type=device.type, enabled=(device.type=="cuda")):
            logits = seg_model(x)
            loss = seg_loss_fn(logits.float(), y)
        seg_scaler.scale(loss).backward()
        seg_scaler.step(seg_opt); seg_scaler.update()
        running += loss.item()
    vd, viou = evaluate_seg(seg_model, seg_val_loader)
    score = float(np.mean(vd))
    seg_sched.step(score)
    seg_history.append({"epoch":epoch, "train_loss":running/max(len(seg_train_loader),1),
                        "val_dice_disc":vd[0], "val_dice_cup":vd[1], "val_mean_dice":score,
                        "val_iou_disc":viou[0], "val_iou_cup":viou[1]})
    if score > best_seg_score:
        best_seg_score, no_improve = score, 0
        torch.save(seg_model.state_dict(), SEG_BEST)
        mark = " * best"
    else:
        no_improve += 1; mark = ""
    print(f"Epoch {epoch:02d}/{SEG_EPOCHS} | loss={running/max(len(seg_train_loader),1):.4f} "
          f"| val disc Dice={vd[0]:.4f} cup Dice={vd[1]:.4f} "
          f"| IoU disc={viou[0]:.4f} cup={viou[1]:.4f}{mark}")
    if no_improve >= patience:
        print("Early stopping."); break

pd.DataFrame(seg_history).to_csv(OUT / "segmentation_training_history.csv", index=False)
seg_model.load_state_dict(torch.load(SEG_BEST, map_location=device))
test_dice, test_iou = evaluate_seg(seg_model, seg_test_loader)
print(f"OFFICIAL REFUGE TEST Dice — disc: {test_dice[0]:.4f} | cup: {test_dice[1]:.4f}")
print(f"OFFICIAL REFUGE TEST IoU  — disc: {test_iou[0]:.4f} | cup: {test_iou[1]:.4f}")

# Show predictions vs ground truth on a few held-out test images.
@torch.no_grad()
def predict_masks(model, frame, size=SEG_SIZE, batch_size=4):
    model.eval(); outputs=[]
    ds = FundusSegDataset(frame, augment=False)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)
    for x,_ in loader:
        x=x.to(device)
        with torch.autocast(device_type=device.type, enabled=(device.type=="cuda")):
            p=torch.sigmoid(model(x))
        pred = (p >= 0.5).cpu().numpy().astype(np.uint8)
        pred[:, 1] = pred[:, 1] * pred[:, 0]
        outputs.extend(pred)
    return np.stack(outputs)

test_pred_masks = predict_masks(seg_model, test_df)
pick = np.linspace(0, len(test_df)-1, min(4,len(test_df)), dtype=int)
fig, axes = plt.subplots(len(pick), 3, figsize=(10, 3.2*len(pick)))
if len(pick)==1: axes=np.expand_dims(axes,0)
for row, i in enumerate(pick):
    img=read_fundus(test_df.iloc[i].image, SEG_SIZE)
    gt=read_disc_cup(test_df.iloc[i]["mask"], SEG_SIZE)
    axes[row,0].imshow(img); axes[row,0].set_title("Fundus")
    axes[row,1].imshow(img); axes[row,1].imshow(gt[0], alpha=.35, cmap="Greens"); axes[row,1].imshow(gt[1], alpha=.45, cmap="Reds"); axes[row,1].set_title("Ground truth")
    axes[row,2].imshow(img); axes[row,2].imshow(test_pred_masks[i,0], alpha=.35, cmap="Greens"); axes[row,2].imshow(test_pred_masks[i,1], alpha=.45, cmap="Reds"); axes[row,2].set_title("Prediction")
    for a in axes[row]: a.axis("off")
plt.tight_layout(); plt.savefig(OUT/"segmentation_examples.png", dpi=140); plt.show()
```

```python

CLS_SIZE = 224
CLS_BATCH = 24
CLS_EPOCHS = 20
CLS_BEST = OUT / "resnet18_glaucoma_best.pth"

class FundusClsDataset(Dataset):
    def __init__(self, frame, augment=False):
        self.frame=frame.reset_index(drop=True); self.augment=augment
    def __len__(self): return len(self.frame)
    def __getitem__(self, idx):
        r=self.frame.iloc[idx]
        x=read_fundus(r.image, CLS_SIZE)
        if self.augment:
            if random.random()<0.5: x=np.ascontiguousarray(x[:,::-1])
            if random.random()<0.5:
                x=np.clip(x.astype(np.float32)*random.uniform(.9,1.1)+random.uniform(-8,8),0,255).astype(np.uint8)
        x=x.astype(np.float32)/255.0
        x=(x-IMAGENET_MEAN)/IMAGENET_STD
        return torch.from_numpy(x.transpose(2,0,1).copy()), torch.tensor(int(r.label),dtype=torch.long)

cls_train_loader=DataLoader(FundusClsDataset(train_df,True),batch_size=CLS_BATCH,shuffle=True,num_workers=2,pin_memory=True)
cls_val_loader=DataLoader(FundusClsDataset(val_df,False),batch_size=CLS_BATCH,shuffle=False,num_workers=2,pin_memory=True)
cls_test_loader=DataLoader(FundusClsDataset(test_df,False),batch_size=CLS_BATCH,shuffle=False,num_workers=2,pin_memory=True)

weights=torchvision.models.ResNet18_Weights.DEFAULT
cls_model=torchvision.models.resnet18(weights=weights)
cls_model.fc=nn.Linear(cls_model.fc.in_features,2)
cls_model=cls_model.to(device)

counts=np.bincount(train_df.label.values,minlength=2)
class_weights=torch.tensor(len(train_df)/(2*np.maximum(counts,1)),dtype=torch.float32,device=device)
cls_criterion=nn.CrossEntropyLoss(weight=class_weights)
cls_opt=torch.optim.AdamW(cls_model.parameters(),lr=1e-4,weight_decay=1e-4)
cls_sched=torch.optim.lr_scheduler.ReduceLROnPlateau(cls_opt,mode="max",factor=.5,patience=2)
cls_scaler=torch.amp.GradScaler("cuda",enabled=(device.type=="cuda"))

@torch.no_grad()
def predict_classifier(model, loader):
    model.eval(); ys=[]; ps=[]
    for x,y in loader:
        x=x.to(device)
        with torch.autocast(device_type=device.type,enabled=(device.type=="cuda")):
            logits=model(x)
        ps.extend(torch.softmax(logits.float(),1)[:,1].cpu().numpy())
        ys.extend(y.numpy())
    return np.asarray(ys),np.asarray(ps)

best_auc=-1; no_improve=0; cls_patience=5; cls_history=[]
for epoch in range(1,CLS_EPOCHS+1):
    cls_model.train(); total=0
    for x,y in cls_train_loader:
        x,y=x.to(device,non_blocking=True),y.to(device,non_blocking=True)
        cls_opt.zero_grad(set_to_none=True)
        with torch.autocast(device_type=device.type,enabled=(device.type=="cuda")):
            logits=cls_model(x); loss=cls_criterion(logits,y)
        cls_scaler.scale(loss).backward(); cls_scaler.step(cls_opt); cls_scaler.update()
        total+=loss.item()
    vy,vp=predict_classifier(cls_model,cls_val_loader)
    auc=float(roc_auc_score(vy,vp)) if len(np.unique(vy))==2 else float("nan")
    cls_sched.step(auc if np.isfinite(auc) else 0.0)
    cls_history.append({"epoch":epoch,"train_loss":total/max(len(cls_train_loader),1),"val_auc":auc})
    if np.isfinite(auc) and auc>best_auc:
        best_auc=auc; no_improve=0; torch.save(cls_model.state_dict(),CLS_BEST); mark=" * best"
    else:
        no_improve+=1; mark=""
    print(f"Epoch {epoch:02d}/{CLS_EPOCHS} | train loss={total/max(len(cls_train_loader),1):.4f} | val AUROC={auc:.4f}{mark}")
    if no_improve>=cls_patience:
        print("Early stopping."); break

pd.DataFrame(cls_history).to_csv(OUT/"classification_training_history.csv",index=False)
cls_model.load_state_dict(torch.load(CLS_BEST,map_location=device))

# Select the operating threshold using REFUGE validation data only.
vy, vp = predict_classifier(cls_model, cls_val_loader)
if len(np.unique(vy)) != 2:
    raise ValueError("Validation set must contain both classes for threshold selection.")
fpr_val, tpr_val, thresholds_val = roc_curve(vy, vp)
valid_thresholds = np.isfinite(thresholds_val)
best_threshold = float(thresholds_val[valid_thresholds][
    np.argmax((tpr_val - fpr_val)[valid_thresholds])
])
print("Selected REFUGE validation threshold:", best_threshold)

# Evaluate the untouched official REFUGE test set with that fixed threshold.
yt,pt=predict_classifier(cls_model,cls_test_loader)
pred=(pt>=best_threshold).astype(int)
tn,fp,fn,tp=confusion_matrix(yt,pred,labels=[0,1]).ravel()
final_metrics={
    "test_n":int(len(yt)),
    "accuracy":float(accuracy_score(yt,pred)),
    "balanced_accuracy":float(balanced_accuracy_score(yt,pred)),
    "auroc":float(roc_auc_score(yt,pt)),
    "sensitivity":float(tp/max(tp+fn,1)),
    "specificity":float(tn/max(tn+fp,1)),
    "precision":float(tp/max(tp+fp,1)),
    "f1_score":float(2*tp/max(2*tp+fp+fn,1)),
    "decision_threshold":best_threshold,
    "threshold_selection_method": "Youden's J on validation set",
    "confusion_matrix_TN_FP_FN_TP":[int(tn),int(fp),int(fn),int(tp)],
    "validation_best_auc":float(best_auc)
}
print("\nFINAL OFFICIAL REFUGE TEST METRICS")
print(json.dumps(final_metrics,indent=2))
print(classification_report(yt,pred,target_names=["Normal","Glaucoma"],zero_division=0))


# Independent external-dataset evaluation (no retraining or threshold tuning).
# The external label parser accepts only explicit class-folder names or CSV manifests.
external_results = {}
for ds_name, ext_df in external_frames.items():
    ext_loader = DataLoader(
        FundusClsDataset(ext_df, augment=False),
        batch_size=CLS_BATCH, shuffle=False, num_workers=2, pin_memory=True
    )
    ey, ep = predict_classifier(cls_model, ext_loader)
    if len(np.unique(ey)) < 2:
        print(f"Skipping AUROC for {ds_name}: only one class was found.")
        ext_auc = None
    else:
        ext_auc = float(roc_auc_score(ey, ep))
    epred = (ep >= best_threshold).astype(int)  # REFUGE validation threshold; no external tuning
    etn, efp, efn, etp = confusion_matrix(ey, epred, labels=[0,1]).ravel()
    external_results[ds_name] = {
        "n": int(len(ey)),
        "auroc": ext_auc,
        "accuracy": float(accuracy_score(ey, epred)),
        "balanced_accuracy": float(balanced_accuracy_score(ey, epred)),
        "sensitivity": float(etp / max(etp + efn, 1)),
        "specificity": float(etn / max(etn + efp, 1)),
        "precision": float(etp / max(etp + efp, 1)),
        "f1_score": float(2 * etp / max(2 * etp + efp + efn, 1)),
        "decision_threshold": best_threshold,
        "confusion_matrix_TN_FP_FN_TP": [int(etn), int(efp), int(efn), int(etp)]
    }
    print(f"\n{ds_name} EXTERNAL RESULTS")
    print(json.dumps(external_results[ds_name], indent=2))

with open(OUT / "external_dataset_metrics.json", "w") as f:
    json.dump(external_results, f, indent=2)
with open(OUT/"classification_test_metrics.json","w") as f: json.dump(final_metrics,f,indent=2)

fpr,tpr,_=roc_curve(yt,pt)
plt.figure(figsize=(5,4)); plt.plot(fpr,tpr,label=f"ResNet18 AUROC={final_metrics['auroc']:.3f}")
plt.plot([0,1],[0,1],"k--"); plt.xlabel("False positive rate"); plt.ylabel("True positive rate")
plt.title("REFUGE official test ROC"); plt.legend(); plt.tight_layout()
plt.savefig(OUT/"classification_roc.png",dpi=140); plt.show()

```

```python

def gradcam_resnet(model, x_tensor, target_class=1):
    model.eval()
    activations = {}
    gradients = {}
    layer = model.layer4[-1]

    def save_activation(module, inputs, output):
        activations["value"] = output
        output.register_hook(lambda grad: gradients.__setitem__("value", grad))

    handle = layer.register_forward_hook(save_activation)
    try:
        model.zero_grad(set_to_none=True)
        logits = model(x_tensor)
        logits[:, target_class].sum().backward()

        acts = activations["value"]
        grads = gradients["value"]
        weights = grads.mean(dim=(2, 3), keepdim=True)
        cam = torch.relu((weights * acts).sum(dim=1, keepdim=True))
        cam = torch.nn.functional.interpolate(
            cam, size=(CLS_SIZE, CLS_SIZE), mode="bilinear", align_corners=False
        )
        cam = cam[0, 0].detach().cpu().numpy()
        cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        return cam, logits.detach()
    finally:
        handle.remove()

def prep_cls_image(path):
    x=read_fundus(path,CLS_SIZE).astype(np.float32)/255.0
    norm=(x-IMAGENET_MEAN)/IMAGENET_STD
    return x,torch.from_numpy(norm.transpose(2,0,1).copy()).unsqueeze(0).to(device)

pred_test=(pt>=best_threshold).astype(int)
candidate=list(np.where(pred_test==yt)[0][:2])+list(np.where(pred_test!=yt)[0][:2])
candidate=list(dict.fromkeys(candidate))[:4]
if candidate:
    fig,axes=plt.subplots(len(candidate),2,figsize=(8,3.5*len(candidate)))
    if len(candidate)==1: axes=np.expand_dims(axes,0)
    for j,i in enumerate(candidate):
        rgb,x=prep_cls_image(test_df.iloc[i].image)
        cam,_=gradcam_resnet(cls_model,x,1)
        axes[j,0].imshow(rgb); axes[j,0].set_title(f"True={yt[i]} | p(glaucoma)={pt[i]:.2f}"); axes[j,0].axis("off")
        axes[j,1].imshow(rgb); axes[j,1].imshow(cam,cmap="jet",alpha=.40); axes[j,1].set_title("Grad-CAM for glaucoma class"); axes[j,1].axis("off")
    plt.tight_layout(); plt.savefig(OUT/"gradcam_examples.png",dpi=140); plt.show()


def largest_component(mask):
    n, lab, stats, centroids = cv2.connectedComponentsWithStats(
        mask.astype(np.uint8),
        connectivity=8
    )
    if n <= 1:
        return np.zeros_like(mask, dtype=bool)
    k = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
    return lab == k

def cdr_from_masks(masks):
    vals=[]
    for disc,cup in zip(masks[:,0],masks[:,1]):
        d=largest_component(disc)
        c=(cup>0)&d
        yd,xd=np.where(d); yc,xc=np.where(c)
        if len(yd)<10 or len(yc)<5:
            vals.append([np.nan,np.nan,np.nan]); continue
        v=(yc.max()-yc.min()+1)/(yd.max()-yd.min()+1)
        h=(xc.max()-xc.min()+1)/(xd.max()-xd.min()+1)
        area=c.sum()/max(d.sum(),1)
        vals.append([v,h,area])
    return np.asarray(vals)

cdr_test=cdr_from_masks(test_pred_masks)
cdr_df=pd.DataFrame({
    "image":test_df.image.values,"true_label":yt,"glaucoma_probability":pt,
    "vertical_CDR":cdr_test[:,0],"horizontal_CDR":cdr_test[:,1],"cup_disc_area_ratio":cdr_test[:,2]
})
cdr_df.to_csv(OUT/"test_predictions_and_CDR.csv",index=False)
print("Mean vertical CDR by REFUGE test label (descriptive only):")
print(cdr_df.groupby("true_label")["vertical_CDR"].mean())

torch.save(seg_model.state_dict(),OUT/"unet_resnet34_best.pth")
torch.save(cls_model.state_dict(),OUT/"resnet18_glaucoma_best.pth")
summary={
    "segmentation_test_dice_disc":float(test_dice[0]),
    "segmentation_test_dice_cup":float(test_dice[1]),
    "segmentation_test_iou_disc":float(test_iou[0]),
    "segmentation_test_iou_cup":float(test_iou[1]),
    "classification":final_metrics,
    "external_classification":external_results,
    "note":"Educational prototype; external validation and clinical review required."
}
with open(OUT/"summary.json","w") as f: json.dump(summary,f,indent=2)
zip_path=shutil.make_archive(str(BASE/"glaucoma_project_results"),"zip",root_dir=OUT)
print("Saved outputs to:",OUT)
print("ZIP:",zip_path)
```
